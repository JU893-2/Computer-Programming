#### we can use an if statement to alter the flow of code. After the word if we include an expression that resolves to true or false. This expression can be built using comparison and logical operatros. If we want one of two things to happen, after out if, we can have an else statement. If there are multiple routes, we can include some elifs in between the if and else.

```python
clothing_stores=["walmart","pandabuy","nike","peter millar","dior"]
fav_store=input("Enter your favorite store: ")
has_store= False
# use a for loop and an if statement to see if their favorite store is in the list of stores
for store in clothing_stores:
    if store.lower() == fav_store.lower():
        has_store=True
if has_store==True:
    print("we have the store")
else:
    print("we don't have it")
```
