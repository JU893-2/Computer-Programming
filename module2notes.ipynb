# i can change the value of an item in the list by resetting it. for example my_list [1] = "new value". I can also use methods on lists. Some methods I can use are .append() and .insert(). Append adds an item to the end of the list. insert adds an item somewhere in the list and shifts everything down to the right. Append takes on parameter which means I put one value in () and that value is the thing I'm adding to the list. Insert takes two parameters. The first is the location, the second is the item to insert at that location.

```python
activities = ["gaming","studying", "eating","resting","running","sleeping"]
print(activities)

#activities.remove("sleeping")
activities.pop(5)
print(activities)

#remove requires value
#pop removes by index 
```

### pop requires me to know the index of the value i'm trying to remove whereas remove requires me to know the exact value 

```python
numbers = [3,5,8,10,11,28,42,77,101,108,2002]
# without modifying the code above 
print(numbers)
# change the 8 into 1
numbers[2] = 1

# insert a 50 at index 2
numbers.insert(2,50)

# remove the 108 by value 
numbers.remove(108)
# add 100 to the end
numbers.append(100)

#remove 2002 by negative index
numbers.pop(-2)
#remove 11 by postive index
numbers.pop(5)
print(numbers)

#use del to delete the 5
del(numbers[1])
print(numbers)
```
